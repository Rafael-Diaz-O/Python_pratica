```python
texto = "teXTo  raro sIn formato" #texto mal escrito

print("Texto: " , texto)

#Funciones
print(texto.upper()) #Coloca todo el texto en mayuscula
print(texto.lower()) #Coloca todo el texto en minuscula
print(texto.strip) #Elimina el espacio al final del texto

print("Texto sin los espacios repetidos en el medio")
print(texto.replace("  "," ")) # lo que hace es si ve que hay dos espacios lo remplazo por un solo espacio

#El replace sirve para cambiar cualquier carater
print(texto.replace("t","r")) # Aqui cambio las t minusculas por una r


#Aplicando varias funciones a la variables texto
print(texto.lower().strip().replace("  "," "))

texto = texto.lower().strip().replace("  "," ") #En la variable texto se guardan el resultado total de haber aplicado todas las funciones a la variable texto

#Ahora al mostrar texto en pantalla se mostrara con todos los cambios guardados
print(texto)

```
