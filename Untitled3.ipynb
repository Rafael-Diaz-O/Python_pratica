```python
from groq import Groq

client = Groq()
completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
      {
        "role": "system",
        "content": "que es la ia\n"
      }
    ],
    temperature=0.5, #Que tan creativa sera la respuesta, la temperatura va de 0-2, cuanto mas baja sea la temperatura menos creativa sera la respuesta y se ira por la respuesta mas probable
    #max_completion_tokens=8192,#Esta linea de codigo me estaba generando el error
    max_tokens = 1024, #Se cambia por esta linea , lo que hace esot es fijar una cantidad de palabras que tendra la respuesta de la ia
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```

```python
#Etapa 1

lista_preguntas = [
    "¿Cual es la empresa mas grande del mundo?",
    "¿Cual es la capital de Colombia?",
    "¿Cual es el rio mas largo del mundo",
    "¿Cual es el planeta mas grande del sistema solar?",
    "¿Cual es el pais mas grande del mundo?",
    "¿Cual es el animal mas grande del mundo?"

]
```

```python
#Etapa 2

#2.1 Crear el archivo txt
with open("Lista-preguntas.txt","w",encoding="utf-8")as preguntas: #La primera parte es el nombre del archivo y la segunda la w es de escritura queremos que escriva la archivo
#El apodo que le solocamos es archivo es decir nos referiremos a el como archivo , el encnding son los carateres que usa el archivo para ser creado como tildes puntos etc
  for resumen in lista_preguntas:
    preguntas.write(resumen + "\n")
```

```python
#2.2 Leer el archivo

lista_lectura_preguntas= []
with open("Lista-preguntas.txt","r",encoding="utf-8")as preguntas: #La primera parte es el nombre del archivo y la segunda la r es de lectura  queremos que se lea el archivo
#El apodo que le solocamos es archivo es decir nos referiremos a el como archivo , el encnding son los carateres que usa el archivo para ser creado como tildes puntos etc
    for linea in preguntas:
        lista_lectura_preguntas.append(linea.strip())
```

```python
#Etapa 3
def responder_preguntas(lista_preguntas):
  lista_respuestas = []
  for numero,pregunta in enumerate(lista_preguntas): #Esta funcion retorna u numero por cada elemento de la lista, en la variable numero guardo los numeros de las funciones
    respuesta = client.chat.completions.create( #Se coloca asi para que sea un poco mas legible y facil guiranos
        model = "llama-3.3-70b-versatile",
                                                #"llama-3.1-8b-instant",
        messages=[

      {
        "role": "user",
        "content":f"""Responde la pregunta {pregunta} en una sola linea """
      }
    ]
        )
    texto = respuesta.choices[0].message.content #variable encrgada de almacenar la repsuesta
    #lista_resumenes.append(texto) #Aqui almacenamos la respuesta
    #Se pone 3 comillas para hacegurarnos que al meter la variable y se un string y tenga comilla no lo interprete como comillla y sierre el texto asi con 3 comillas python sabe que parte es el texto que le pasamos y cual es lo que queremos que analice
    lista_respuestas.append(f"Respuesta a la pregunta N° {numero + 1}:{texto}")
  return lista_respuestas
```

```python
#4 Crear el archivo csv
with open("respuesta.csv","w",encoding="utf-8")as respuestas: #La primera parte es el nombre del archivo y la segunda la w es de escritura queremos que escriva la archivo
#El apodo que le solocamos es archivo es decir nos referiremos a el como archivo , el encnding son los carateres que usa el archivo para ser creado como tildes puntos etc
  for resumen in respuesta:
    respuestas.write(resumen + "\n")
```
